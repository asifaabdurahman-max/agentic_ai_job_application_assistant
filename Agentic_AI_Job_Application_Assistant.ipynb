```python

import re

def analyze_job_description(job_description):
    skills_list = [
        "Python", "SQL", "Excel", "Power BI", "Machine Learning",
        "Deep Learning", "Generative AI", "Agentic AI",
        "Prompt Engineering", "Data Analysis", "Pandas",
        "Communication", "Problem Solving", "Git", "APIs"
    ]

    found_skills = []

    for skill in skills_list:
        pattern = r"\b" + re.escape(skill) + r"\b"
        if re.search(pattern, job_description, re.IGNORECASE):
            found_skills.append(skill)

    return found_skills


job_description = """
We are looking for an AI intern with Python, SQL,
Generative AI, Prompt Engineering, and problem-solving skills.
"""

required_skills = analyze_job_description(job_description)

print("JOB DESCRIPTION ANALYSIS")
print("-------------------------")
print("Skills identified:")

for skill in required_skills:
    print("-", skill)

print("\nTotal skills identified:", len(required_skills))
```

```python

def analyze_resume(resume_text):
    resume_skills = analyze_job_description(resume_text)
    return resume_skills


resume_text = """
MCA graduate with knowledge of Python, SQL, Pandas,
Excel, Generative AI, and Prompt Engineering.
Completed an AI research assistant project using Python.
"""

candidate_skills = analyze_resume(resume_text)

print("RESUME ANALYSIS")
print("----------------")
print("Skills found in resume:")

for skill in candidate_skills:
    print("-", skill)

print("\nTotal skills found:", len(candidate_skills))
```

```python

def match_skills(required_skills, candidate_skills):
    matched_skills = []
    missing_skills = []

    for skill in required_skills:
        if skill.lower() in [s.lower() for s in candidate_skills]:
            matched_skills.append(skill)
        else:
            missing_skills.append(skill)

    if len(required_skills) > 0:
        match_percentage = round(
            len(matched_skills) / len(required_skills) * 100, 2
        )
    else:
        match_percentage = 0

    return {
        "matched_skills": matched_skills,
        "missing_skills": missing_skills,
        "match_percentage": match_percentage
    }


result = match_skills(required_skills, candidate_skills)

print("JOB SKILL MATCH REPORT")
print("----------------------")
print("Matched skills:", result["matched_skills"])
print("Missing skills:", result["missing_skills"])
print("Skill match:", str(result["match_percentage"]) + "%")
```

```python

def generate_interview_questions(missing_skills):
    questions = {}

    for skill in missing_skills:
        questions[skill] = [
            f"What is {skill}, and why is it useful?",
            f"Give a practical example of using {skill}.",
            f"How would you solve a real-world problem using {skill}?"
        ]

    return questions


interview_questions = generate_interview_questions(
    result["missing_skills"]
)

print("INTERVIEW PREPARATION PLAN")
print("==========================")

for skill, questions in interview_questions.items():
    print(f"\nSkill: {skill}")
    for number, question in enumerate(questions, start=1):
        print(f"{number}. {question}")
```

```python

def run_job_assistant(job_description, resume_text):
    print("=" * 45)
    print("AGENTIC AI JOB APPLICATION ASSISTANT")
    print("=" * 45)

    # Step 1: Analyze job requirements
    required = analyze_job_description(job_description)

    # Step 2: Analyze candidate resume
    candidate = analyze_resume(resume_text)

    # Step 3: Match skills
    match_result = match_skills(required, candidate)

    print("\n1. REQUIRED SKILLS")
    print(required)

    print("\n2. CANDIDATE SKILLS")
    print(candidate)

    print("\n3. MATCHED SKILLS")
    print(match_result["matched_skills"])

    print("\n4. MISSING SKILLS")
    print(match_result["missing_skills"])

    print("\n5. SKILL MATCH")
    print(str(match_result["match_percentage"]) + "%")

    # Step 4: Create learning plan
    plan = create_learning_plan(match_result["missing_skills"])

    print("\n6. LEARNING PLAN")
    if plan:
        for skill, action in plan.items():
            print(f"- {skill}: {action}")
    else:
        print("No skill gaps identified.")

    # Step 5: Generate interview questions
    questions = generate_interview_questions(
        match_result["missing_skills"]
    )

    print("\n7. INTERVIEW PREPARATION")
    if questions:
        for skill, items in questions.items():
            print(f"\n{skill}:")
            for question in items:
                print("-", question)
    else:
        print("No missing-skill questions to generate.")

    return {
        "required_skills": required,
        "candidate_skills": candidate,
        "match_result": match_result,
        "learning_plan": plan,
        "interview_questions": questions
    }


print("Complete workflow is ready!")
```

```python

from google.colab import files

def export_job_report(project_result):
    match = project_result["match_result"]

    report = [
        "AGENTIC AI JOB APPLICATION ASSISTANT",
        "=" * 40,
        "",
        f"Skill match: {match['match_percentage']}%",
        "",
        "MATCHED SKILLS:",
        *[f"- {s}" for s in match["matched_skills"]],
        "",
        "MISSING SKILLS:",
        *[f"- {s}" for s in match["missing_skills"]],
        "",
        "LEARNING PLAN:"
    ]

    for skill, action in project_result["learning_plan"].items():
        report.append(f"- {skill}: {action}")

    report.extend(["", "INTERVIEW QUESTIONS:"])

    for skill, questions in project_result["interview_questions"].items():
        report.append(f"\n{skill}:")
        report.extend([f"- {q}" for q in questions])

    filename = "job_application_analysis.txt"

    with open(filename, "w", encoding="utf-8") as f:
        f.write("\n".join(report))

    print("Report created:", filename)
    files.download(filename)


export_job_report(project_result)
```
